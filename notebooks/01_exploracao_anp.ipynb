```python
import pandas as pd
```

```python
arquivo = r"../data/raw/Preços semestrais - AUTOMOTIVOS_2026.01.csv"

df = pd.read_csv(
    arquivo,
    sep=";",
    encoding="utf-8"
)

df.head()
```

```python
print("Linhas:", df.shape[0])
print("Colunas:", df.shape[1])
```

```python
df.info()
```

```python
df.columns.tolist()
```

```python
# Quantidade e percentual de valores ausentes por coluna

valores_ausentes = pd.DataFrame({
    "quantidade_ausente": df.isna().sum(),
    "percentual_ausente": (df.isna().mean() * 100).round(2)
})

valores_ausentes.sort_values(
    by="quantidade_ausente",
    ascending=False
)
```

```python
# Quantidade de linhas completamente duplicadas

quantidade_duplicadas = df.duplicated().sum()

print("Linhas duplicadas:", quantidade_duplicadas)
print(
    "Percentual de duplicadas:",
    round(quantidade_duplicadas / len(df) * 100, 2),
    "%"
)
```

```python
# Valores distintos das principais colunas categóricas

print("Regiões:", sorted(df["Regiao - Sigla"].dropna().unique()))
print("Estados:", df["Estado - Sigla"].nunique())
print("Municípios:", df["Municipio"].nunique())
print("Revendas:", df["CNPJ da Revenda"].nunique())
print("Produtos:", sorted(df["Produto"].dropna().unique()))
print("Bandeiras:", df["Bandeira"].nunique())
```

Também foram encontradas pequenas quantidades de valores ausentes nas colunas `Bairro`, `Numero Rua` e `Revenda`.
